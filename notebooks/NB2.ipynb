```python
# Notebook 2: A/B test cổ điển

import numpy as np
from statsmodels.stats.proportion import proportions_ztest

from hillstrom_uplift.data import load

df = load()


def ab_test(df, group, outcome="visit"):
    t = df.loc[df.treatment == group, outcome]
    c = df.loc[df.treatment == "control", outcome]
    count = np.array([t.sum(), c.sum()])
    nobs = np.array([len(t), len(c)])
    z, p = proportions_ztest(count, nobs)
    lift = t.mean() - c.mean()
    return {
        "group": group,
        "rate_t": t.mean(),
        "rate_c": c.mean(),
        "abs_lift": lift,
        "rel_lift": lift / c.mean(),
        "p": p,
    }


for g in ["mens", "womens"]:
    print(ab_test(df, g, "visit"))
    print(ab_test(df, g, "conversion"))
```
