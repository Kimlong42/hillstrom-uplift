```python
# Notebook 1: EDA + kiểm tra randomization

import pandas as pd
from scipy import stats

from hillstrom_uplift.data import load

df = load()

# 1. Tỷ lệ nhóm ~ 1/3 mỗi nhóm?
print(df["treatment"].value_counts(normalize=True))
```

```python
# 2. Balance biến liên tục: SMD (standardized mean difference)
def smd(a, b):
    return (a.mean() - b.mean()) / ((a.var() + b.var()) / 2) ** 0.5


for col in ["recency", "history"]:
    c = df.loc[df.treatment == "control", col]
    for g in ["mens", "womens"]:
        t = df.loc[df.treatment == g, col]
        print(col, g, round(smd(t, c), 4))  # |SMD| < 0.1 là cân bằng
```

```python
# 3. Balance biến phân loại: chi-square độc lập với treatment
for col in ["zip_code", "channel", "newbie", "mens", "womens"]:
    ct = pd.crosstab(df[col], df["treatment"])
    _, p, _, _ = stats.chi2_contingency(ct)
    print(col, "p =", round(p, 3))
```
